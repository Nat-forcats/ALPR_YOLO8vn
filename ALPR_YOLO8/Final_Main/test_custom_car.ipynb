```python
from ultralytics import YOLO

def main():
    # 1. Load your trained weights
    model = YOLO(r'C:/Users/Tyler/Documents/Schoolwork/IST 691/Final_Main/runs/detect/IST691_Paper/baseline_nano-2/weights/best.pt')

    # 2. Put both image paths into a Python list
    images = ['TylerCar.jpg', 'KimiCar.jpg']

    # 3. Run batch inference
    print(f"Batch analyzing {len(images)} custom vehicles...")
    results = model.predict(
        source=images,
        conf=0.4,       # 40% confidence threshold
        save=True,      # Save the annotated images to disk
        show=True       # Pop open windows showing the boxes
    )
    
    print("Batch inference complete!")

if __name__ == '__main__':
    main()
```
