```python
import torch
from ultralytics import YOLO

def main():
    # 1. Hardware Verification
    if not torch.cuda.is_available():
        print("Error: CUDA not detected. PyTorch is running on CPU.")
        return
        
    device = 0
    print(f"Hardware Verified. Training on: {torch.cuda.get_device_name(device)}")

    # 2. Initialize the Baseline CNN (Nano Scale)
    model = YOLO('yolov8n.pt')

    # 3. Execute Training Loop
    print("Initializing baseline training for IEEE paper...")
    model.train(
        data='plate_data.yaml',
        epochs=30,             # 30 epochs establishes a solid baseline for the report
        imgsz=640,             # Standard resolution for bounding box detection
        batch=32,              # 3080 Ti can easily handle batch 32 at 640px
        workers=8,             # Adjust to your CPU core count for faster data loading
        amp=True,              # Crucial: Activates 3080 Ti Tensor Cores (FP16)
        device=device,
        project='IST691_Paper',# Organizes output folders neatly
        name='baseline_nano'   # Names this specific training run
    )

if __name__ == '__main__':
    main()
```
